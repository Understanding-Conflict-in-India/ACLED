```python
import pandas as pd
import re

# File path
file_path = "/Users/mukundsaraf/Desktop/ACLED data/ACLED Data_2025-12-27.csv"

# Load data
df = pd.read_csv(file_path)

# Detect source columns
source_cols = [col for col in df.columns if "source" in col.lower()]

print("Detected source-related columns:")
print(source_cols)

# Collect unique sources correctly
all_sources = set()

for col in source_cols:
    for cell in df[col].dropna():
        # Split on common ACLED separators (; , |)
        sources = re.split(r"[;,|]", str(cell))
        for s in sources:
            s_clean = s.strip()
            if s_clean:
                all_sources.add(s_clean)

# Convert to sorted list
all_sources = sorted(all_sources)

print("\nList of all UNIQUE source names:")
print(all_sources)

print(f"\nTotal number of unique sources: {len(all_sources)}")
```

```python
import pandas as pd
import re
from collections import Counter

# File path
file_path = "/Users/mukundsaraf/Desktop/ACLED data/ACLED Data_2025-12-27.csv"

# Load data
# Try reading as CSV first as ACLED files often have .xls extension but are CSVs
try:
    df = pd.read_excel(file_path)
except:
    df = pd.read_csv(file_path)

# 1. Collect all individual sources
all_sources = []
if 'source' in df.columns:
    for cell in df['source'].dropna():
        # Split by common delimiters (; | ,) and strip whitespace
        # ACLED sources are typically separated by semicolons
        parts = re.split(r'[;|]', str(cell))
        for p in parts:
            clean_source = p.strip()
            if clean_source:
                all_sources.append(clean_source)

# 2. Count frequencies
source_counts = Counter(all_sources)

# 3. Get Top 3
top_3 = source_counts.most_common(10)

print("Top 10 Sources:")
for source, count in top_3:
    print(f"- {source}: {count}")
```

```python
import pandas as pd

# File path
file_path = "/Users/mukundsaraf/Desktop/ACLED data/ACLED Data_2025-12-27.csv"

# Load data
df = pd.read_csv(file_path)

# Detect conflict-type related columns
conflict_cols = [
    col for col in df.columns
    if any(k in col.lower() for k in ["event_type", "sub_event", "conflict", "interaction"])
]

print("Detected conflict-related columns:")
print(conflict_cols)

# Collect unique conflict types
conflict_types = {}

for col in conflict_cols:
    unique_vals = sorted(df[col].dropna().unique())
    conflict_types[col] = unique_vals

# Print results clearly
for col, vals in conflict_types.items():
    print(f"\nUnique values in '{col}':")
    for v in vals:
        print(f"- {v}")

# Optional: save to Excel
output_path = "/Users/mukundsaraf/Desktop/ACLED data/conflict_types_ACLED.xlsx"
with pd.ExcelWriter(output_path) as writer:
    for col, vals in conflict_types.items():
        pd.DataFrame({col: vals}).to_excel(writer, sheet_name=col[:31], index=False)

print(f"\nConflict types saved to: {output_path}")
```

```python
import pandas as pd

# File path
file_path = "/Users/mukundsaraf/Desktop/ACLED data/ACLED Data_2025-12-27.csv"

# Load data
df = pd.read_csv(file_path)

# -------------------------------
# Use ACLED's canonical date column
# -------------------------------
if "event_date" not in df.columns:
    raise ValueError("event_date column not found — cannot determine temporal coverage")

# Parse event_date safely
df["event_date_parsed"] = pd.to_datetime(df["event_date"], errors="coerce")

# Extract year
df["year"] = df["event_date_parsed"].dt.year

# Drop missing
years = sorted(df["year"].dropna().astype(int).unique())

# Temporal coverage
start_year = min(years)
end_year = max(years)

# Missing years
full_range = set(range(start_year, end_year + 1))
missing_years = sorted(full_range - set(years))

# -------------------------------
# Print NOTE-READY summary
# -------------------------------
print("TEMPORAL COVERAGE SUMMARY\n")
print(f"Start year: {start_year}")
print(f"End year: {end_year}")
print("Data frequency: Event-level (multiple events per year)")

if missing_years:
    print(f"Missing years: {missing_years}")
else:
    print("Missing years: None (continuous annual coverage)")
```
